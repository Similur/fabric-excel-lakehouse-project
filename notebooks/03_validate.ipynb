# 03: Validate the Silver output

Check that Silver contains rows, has one row per record ID, and has no blank IDs or invalid amounts before using it for reporting. Print the Bronze and Silver row counts to compare the ingestion history with the current records.

```python
from pyspark.sql import functions as F

bronze = spark.table('demo_bronze_activity')
silver = spark.table('demo_silver_activity')

bronze_rows = bronze.count()
silver_rows = silver.count()
distinct_ids = silver.select('record_id').distinct().count()
missing_ids = silver.filter(F.col('record_id').isNull() | (F.col('record_id') == '')).count()
invalid_amounts = silver.filter(F.col('amount').isNull()).count()

print(f'Bronze history rows: {bronze_rows}')
print(f'Silver current rows: {silver_rows}')
print(f'Distinct Silver IDs: {distinct_ids}')
assert silver_rows == distinct_ids, 'Silver contains duplicate record IDs'
assert missing_ids == 0, 'Silver contains empty record IDs'
assert invalid_amounts == 0, 'Silver contains invalid amounts'
assert silver_rows > 0, 'Silver table is empty'
print('Validation checks passed')
```
